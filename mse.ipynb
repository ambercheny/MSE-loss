```python
import numpy as np
import pandas as pd
```

```python
def interpret_mse(y_true, y_pred, target_std=None, standardized=False):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    mse = np.mean((y_pred - y_true) ** 2)
    rmse = np.sqrt(mse)

    baseline_pred = np.full_like(y_true, y_true.mean(), dtype=float)
    baseline_mse = np.mean((baseline_pred - y_true) ** 2)
    print(f"baseline_pred: {baseline_pred}")
    print(f"baseline_mse: {baseline_mse}")

    relative_mse = mse / baseline_mse if baseline_mse != 0 else np.nan
    r2 = 1 - relative_mse if baseline_mse != 0 else np.nan

    result = {
        "mse": mse,
        "rmse": rmse,
        "baseline_mse_mean_predictor": baseline_mse,
        "relative_mse": relative_mse,
        "r2": r2,
    }

    if standardized:
        result["interpretation"] = f"RMSE is about {rmse:.3f} standard deviations."
    elif target_std is not None:
        result["rmse_in_std_units"] = rmse / target_std
        result["interpretation"] = (
            f"RMSE is {rmse:.3f} original units, "
            f"or {rmse / target_std:.3f} standard deviations."
        )

    return result


def print_mse_report(y_true, y_pred, standardized=False, target_std=None):
    print(f"standardized: {standardized}")
    print(f"y_true: {y_true}")
    print(f"y_pred: {y_pred}")
    metrics = interpret_mse(
        y_true,
        y_pred,
        standardized=standardized,
        target_std=target_std,
    )

    print(f"MSE: {metrics['mse']:.4f}")
    print(f"RMSE: {metrics['rmse']:.4f}")
    print(f"Mean-baseline MSE: {metrics['baseline_mse_mean_predictor']:.4f}")
    print(f"Relative MSE: {metrics['relative_mse']:.4f}")
    print(f"R^2: {metrics['r2']:.4f}")

    if "interpretation" in metrics:
        print(metrics["interpretation"])
```

```python
print_mse_report(y_true = [0.3, 10], y_pred = [0, -0.2], standardized=True)
```
